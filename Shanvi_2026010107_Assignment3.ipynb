Question 1

```python
import seaborn as sns
import matplotlib.pyplot as plt
titanic = sns.load_dataset("titanic")
print(titanic.head())
print(titanic.info())
sns.scatterplot(
    data=titanic,
    x="age",
    y="fare"
)
plt.title("Titanic - Age vs Fare")
plt.show()
sns.scatterplot(
    data=titanic,
    x="age",
    y="fare",
    hue="sex"
)
plt.title("Titanic - Age vs Fare by Sex")
plt.show()
sns.scatterplot(
    data=titanic,
    x="age",
    y="fare",
    hue="sex",
    size="pclass"
)
plt.title("Titanic - Age vs Fare")
plt.show()
sns.lineplot(
    data=titanic,
    x="age",
    y="fare"
)
plt.title("Titanic - Age vs Fare Line Plot")
plt.show()
sns.lineplot(
    data=titanic,
    x="age",
    y="fare",
    hue="sex"
)
plt.title("Titanic - Age vs Fare by Sex")
plt.show()
sns.relplot(
    data=titanic,
    x="age",
    y="fare",
    kind="scatter"
)
plt.show()
sns.relplot(
    data=titanic,
    x="age",
    y="fare",
    kind="line"
)
plt.show()
sns.relplot(
    data=titanic,
    x="age",
    y="fare",
    hue="sex",
    kind="scatter"
)
plt.show()
sns.relplot(
    data=titanic,
    x="age",
    y="fare",
    hue="sex",
    row="survived",
    kind="scatter"
)
plt.show()
sns.relplot(
    data=titanic,
    x="age",
    y="fare",
    hue="sex",
    col="pclass",
    kind="scatter"
)
plt.show()
```

Question 2

```python
import seaborn as sns
import matplotlib.pyplot as plt
flights = sns.load_dataset("flights")
print(flights.head())
print(flights.info())
sns.scatterplot(
    data=flights,
    x="year",
    y="passengers"
)
plt.title("Flights - Year vs Passengers")
plt.show()
sns.scatterplot(
    data=flights,
    x="year",
    y="passengers",
    hue="month"
)
plt.title("Flights - Passengers by Year and Month")
plt.show()
sns.lineplot(
    data=flights,
    x="year",
    y="passengers"
)
plt.title("Flights - Passenger Trend")
plt.show()
sns.lineplot(
    data=flights,
    x="year",
    y="passengers",
    hue="month"
)
plt.title("Flights - Passenger Trend by Month")
plt.show()
sns.relplot(
    data=flights,
    x="year",
    y="passengers",
    kind="scatter"
)
plt.show()
sns.relplot(
    data=flights,
    x="year",
    y="passengers",
    kind="line"
)
plt.show()
sns.relplot(
    data=flights,
    x="year",
    y="passengers",
    hue="month",
    kind="line"
)
plt.show()
sns.relplot(
    data=flights,
    x="year",
    y="passengers",
    col="month",
    col_wrap=4,
    kind="line"
)
plt.show()
sns.relplot(
    data=flights,
    x="year",
    y="passengers",
    row="month",
    kind="line"
)
plt.show()
sns.relplot(
    data=flights,
    x="year",
    y="passengers",
    hue="month",
    col="month",
    col_wrap=4,
    kind="line"
)
plt.show()
```
