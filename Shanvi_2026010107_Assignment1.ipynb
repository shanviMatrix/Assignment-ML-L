Question 1

```python
np.random.seed(42)

# Create 10x10 random array
arr = np.random.randint(1, 100, size=(10, 10))

print("Original 10x10 array:")
print(arr)

# Slicing
A = arr[0:5, 0:7]      # 5x7
B = arr[0:7, 5:10]     # 7x5
C = arr[5:10, 0:7]     # 5x7

print("\nA =")
print(A)

print("\nB =")
print(B)

print("\nC =")
print(C)

# Sum of A and C
sum_AC = A + C

print("\nA + C =")
print(sum_AC)

# Matrix multiplication of A and B
multiplication_AB = A @ B

print("\nA x B =")
print(multiplication_AB)
```

Question 3

```python
np.random.seed(42)

arr = np.random.randint(10, 99, size=(6, 6))

print("Original array:")
print(arr)
```

3 (a)

```python
row_sum = np.sum(arr, axis=1)

print("Sum of each row:")
print(row_sum)
```

```python
column_sum = np.sum(arr, axis=0)

print("Sum of each column:")
print(column_sum)
```

3 (b)

```python
row_mean = np.mean(arr, axis=1)

print("Mean of each row:")
print(np.round(row_mean, 2))
```

3(c)

```python
max_values = np.max(arr, axis=1)
max_indices = np.argmax(arr, axis=1)

print("Maximum value in each row:")
print(max_values)

print("Column index of maximum value:")
print(max_indices)
```

3 (d)

```python
float_arr = arr.astype(float)

print("Float array:")
print(float_arr)
```

Question 4

```python
arr = np.arange(1, 21).reshape(4, 5)

print("4x5 array:")
print(arr)
```

Sub Array

```python
sub_array = arr[1:3, 2:5]

print("Extracted sub-array:")
print(sub_array)
```

Variance

```python
variance = np.var(sub_array)

print("Variance:", variance)
```

Standard Deviation

```python
std = np.std(sub_array)

print("Standard deviation:", std)
```

Variance and Standard Deviation (rounded off)

```python
print("Variance:", round(variance, 2))
print("Standard deviation:", round(std, 2))
```
