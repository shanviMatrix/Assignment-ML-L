## 13. Covariance and correlation between rows (RK2 example)

```python
s1 = pd.Series(data.iloc[0, :4])
s2 = pd.Series(data.iloc[2, :4])

print("Covariance:", s1.cov(s2))
print("Correlation:", s1.corr(s2))
```

## 14. Required covariance and correlation

Numeric-label mapping:

- **1 = sepal.length**
- **2 = sepal.width**
- **3 = petal.length**
- **4 = petal.width**

Find covariance and correlation for **1&2, 1&3, and 1&4**.


```python
# 1 & 2
d1 = data["sepal.length"]
d2 = data["sepal.width"]

print("Covariance (1,2):", d1.cov(d2))
print("Correlation (1,2):", d1.corr(d2))
```

```python
# 1 & 3
d1 = data["sepal.length"]
d3 = data["petal.length"]

print("Covariance (1,3):", d1.cov(d3))
print("Correlation (1,3):", d1.corr(d3))
```

```python
# 1 & 4
d1 = data["sepal.length"]
d4 = data["petal.width"]

print("Covariance (1,4):", d1.cov(d4))
print("Correlation (1,4):", d1.corr(d4))
```

## 15. Mean and standard deviation for labels 1–4

```python
numeric_cols = ["sepal.length", "sepal.width", "petal.length", "petal.width"]

means = data[numeric_cols].mean()
stds = data[numeric_cols].std()

print("Mean:")
print(means)

print("\nStandard deviation:")
print(stds)
```

## 16. Match mean and standard deviation with describe()

```python
comparison = pd.DataFrame({
    "Mean": data[numeric_cols].mean(),
    "Std": data[numeric_cols].std(),
    "Describe_Mean": data.describe().loc["mean"],
    "Describe_Std": data.describe().loc["std"]
})

comparison["Mean_Match"] = comparison["Mean"].round(10) == comparison["Describe_Mean"].round(10)
comparison["Std_Match"] = comparison["Std"].round(10) == comparison["Describe_Std"].round(10)

comparison
```

## 17. Final required results

```python
print("COVARIANCE")
print("1 & 2 =", data["sepal.length"].cov(data["sepal.width"]))
print("1 & 3 =", data["sepal.length"].cov(data["petal.length"]))
print("1 & 4 =", data["sepal.length"].cov(data["petal.width"]))

print("\nCORRELATION")
print("1 & 2 =", data["sepal.length"].corr(data["sepal.width"]))
print("1 & 3 =", data["sepal.length"].corr(data["petal.length"]))
print("1 & 4 =", data["sepal.length"].corr(data["petal.width"]))

print("\nMEAN")
print(data[numeric_cols].mean())

print("\nSTANDARD DEVIATION")
print(data[numeric_cols].std())
```
